# Submission Notebook  
This notebook loads pre-trained models saved from the modeling notebook and generates Kaggle submission files.
No training happens here.

### Load Processed Train & Test Data

```python
import pandas as pd

train = pd.read_csv("../data/processed/processed_train.csv")
test = pd.read_csv("../data/processed/processed_test.csv")

train_ids = train["ProfileID"]
test_ids  = test["ProfileID"]

X_train_full = train.drop(["RiskFlag"], axis=1)
y_train_full = train["RiskFlag"]
X_test_full  = test.drop(["ProfileID"], axis=1)

print("Loaded processed train and test.")
```

### Load all models

```python
import joblib

log_reg = joblib.load("../models/logistic.pkl")
log_reg_bal = joblib.load("../models/logistic_balanced.pkl")
rbf_fast_svm = joblib.load("../models/rbf_fast_svm.pkl")
sigmoid_svm = joblib.load("../models/sigmoid_svm.pkl")
rbf_tuned = joblib.load("../models/rbf_tuned_svm.pkl")
nn_baseline = joblib.load("../models/nn_baseline.pkl")
nn_lbfgs = joblib.load("../models/nn_lbfgs.pkl")
nn_smote = joblib.load("../models/nn_smote.pkl")

print("All models loaded.")
```

### Generate Submission Files

```python
import os
os.makedirs("../submissions/", exist_ok=True)

def save_submission(model, name):
    preds = model.predict(X_test_full)
    sub = pd.DataFrame({
        "ProfileID": test_ids,
        "RiskFlag": preds
    })
    path = f"../submissions/{name}.csv"
    sub.to_csv(path, index=False)
    print("Saved:", path)

save_submission(log_reg, "sub_logistic")
save_submission(log_reg_bal, "sub_logistic_balanced")
save_submission(rbf_fast_svm, "sub_rbf_fast")
save_submission(sigmoid_svm, "sub_sigmoid")
save_submission(rbf_tuned, "sub_rbf_tuned")
save_submission(nn_baseline, "sub_nn_baseline")
save_submission(nn_lbfgs, "sub_nn_lbfgs")
save_submission(nn_smote, "sub_nn_smote")
```
