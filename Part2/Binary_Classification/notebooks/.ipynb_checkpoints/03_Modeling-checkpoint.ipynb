# 03 — Modeling (Logistic Regression, SVM, Neural Network)

This notebook loads the processed dataset from `02_Preprocessing.ipynb`
and trains multiple ML models:

- Logistic Regression  
- SVM (Approx RBF via RBFSampler)  
- FAISS KNN classifier  
- Tuned RBF SVM  
- Neural Networks (Adam, LBFGS, SMOTE + threshold tuning)

We evaluate all models on the same train-test split.

### Imports + Load Processed Train

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, f1_score, recall_score, precision_score
import time
from sklearn.svm import SVC
from sklearn.kernel_approximation import RBFSampler
from sklearn.pipeline import Pipeline
from sklearn.svm import LinearSVC
from sklearn.neural_network import MLPClassifier
from imblearn.over_sampling import RandomOverSampler
from imblearn.over_sampling import SMOTE
import joblib
import os

# Load processed dataset
df = pd.read_csv("../data/processed/processed_full/processed_train.csv")

df.head()
```

### Train/Test Split + Scaling Numerical Columns

```python
results =[]
# Separate features and target
X = df.drop("RiskFlag", axis=1)
y = df["RiskFlag"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Identify numerical columns for scaling
numerical_cols = [
    "ApplicantYears", "AnnualEarnings", "RequestedSum", "TrustMetric",
    "WorkDuration", "ActiveAccounts", "OfferRate", "RepayPeriod", "DebtFactor"
]

# Scale numeric features only
scaler = StandardScaler()
X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])

X_train.head()
```

### Logistic Regression (baseline)

```python
# Train Logistic Regression
start = time.time()
log_reg = LogisticRegression(max_iter=500)
log_reg.fit(X_train, y_train)
train_time = time.time() - start

# Predictions
log_preds = log_reg.predict(X_test)

# Metrics
print("Accuracy:", accuracy_score(y_test, log_preds))
print("\nClassification Report:")
print(classification_report(y_test, log_preds))

# Logging results
results.append({
    "Model": "Logistic Regression",
    "Accuracy": accuracy_score(y_test, log_preds),
    "F1_Positive": f1_score(y_test, log_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, log_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, log_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Logistic Regression (class_weight = 'balanced')

```python
print("Training Logistic Regression (Balanced)...")

# Train Logistic Regression (balanced)
start = time.time()
log_reg_bal = LogisticRegression(class_weight="balanced", max_iter=500)
log_reg_bal.fit(X_train, y_train)
train_time = time.time() - start

# Predictions
log_preds_bal = log_reg_bal.predict(X_test)

# Metrics
print("Accuracy:", accuracy_score(y_test, log_preds_bal))
print("\nClassification Report:")
print(classification_report(y_test, log_preds_bal))

# Logging result
results.append({
    "Model": "Logistic Regression (Balanced)",
    "Accuracy": accuracy_score(y_test, log_preds_bal),
    "F1_Positive": f1_score(y_test, log_preds_bal, pos_label=1),
    "Recall_Positive": recall_score(y_test, log_preds_bal, pos_label=1),
    "Precision_Positive": precision_score(y_test, log_preds_bal, pos_label=1),
    "Training_Time": train_time
})
```

### Linear SVM

```python
### Linear SVM (LinearSVC)

from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, f1_score, recall_score, precision_score
import time

print("Training Linear SVM...")

# Train LinearSVC
start = time.time()
lin_svm = LinearSVC(class_weight='balanced', max_iter=5000, random_state=42)
lin_svm.fit(X_train, y_train)
train_time = time.time() - start

# Predictions
lin_preds = lin_svm.predict(X_test)

print("Training time:", train_time)
print("Accuracy:", accuracy_score(y_test, lin_preds))
print("\nClassification Report:")
print(classification_report(y_test, lin_preds))

# Log results
results.append({
    "Model": "Linear SVM",
    "Accuracy": accuracy_score(y_test, lin_preds),
    "F1_Positive": f1_score(y_test, lin_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, lin_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, lin_preds, pos_label=1),
    "Training_Time": train_time
})
```

### RBF SVM(Standard SVC)

```python
print("Training RBF SVM")

start = time.time()

rbf_svm = SVC(
    kernel='rbf',
    C=1.0,
    gamma='scale',
    class_weight='balanced'
)

rbf_svm.fit(X_train, y_train)

train_time = time.time() - start

# Predictions
rbf_preds = rbf_svm.predict(X_test)

print("Training time:", train_time)
print("Accuracy:", accuracy_score(y_test, rbf_preds))
print("\nClassification Report:")
print(classification_report(y_test, rbf_preds))

# Log results
results.append({
    "Model": "RBF SVM",
    "Accuracy": accuracy_score(y_test, rbf_preds),
    "F1_Positive": f1_score(y_test, rbf_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, rbf_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, rbf_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Fast RBF SVM (RBFSampler + LinearSVC)

```python
print("Training Fast RBF SVM (Fourier Features)...")

# Convert to float32 to reduce memory usage
X_train_32 = X_train.astype(np.float32)
X_test_32  = X_test.astype(np.float32)

start = time.time()

# gamma = 1 / number of features (safe baseline)
gamma_value = 1 / X_train_32.shape[1]
print("Using gamma =", gamma_value)

# RBFSampler (Random Fourier Features)
feature_map_rff = RBFSampler(
    gamma=gamma_value,
    n_components=500,      # safe + fast (you tested before)
    random_state=42
)

# Linear SVM on transformed features
fast_rbf_svm = Pipeline([
    ('rff', feature_map_rff),
    ('svm', LinearSVC(class_weight='balanced', max_iter=5000))
])

fast_rbf_svm.fit(X_train_32, y_train)

train_time = time.time() - start

# Predictions
rff_preds = fast_rbf_svm.predict(X_test_32)

print("Training Time:", train_time)
print("Accuracy:", accuracy_score(y_test, rff_preds))
print("\nClassification Report:")
print(classification_report(y_test, rff_preds))

# Log results
results.append({
    "Model": "Fast RBF SVM (Fourier)",
    "Accuracy": accuracy_score(y_test, rff_preds),
    "F1_Positive": f1_score(y_test, rff_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, rff_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, rff_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Sigmoid SVM

```python
print("Training Sigmoid SVM...")

start = time.time()

sigmoid_svm = SVC(
    kernel="sigmoid",
    class_weight="balanced",
    gamma="scale",
    C=1.0,
    probability=False,   # faster
    max_iter=5000
)

sigmoid_svm.fit(X_train, y_train)

train_time = time.time() - start

sigmoid_preds = sigmoid_svm.predict(X_test)

print("Training Time:", train_time)
print("Accuracy:", accuracy_score(y_test, sigmoid_preds))
print("\nClassification Report:")
print(classification_report(y_test, sigmoid_preds))

# Log results
results.append({
    "Model": "Sigmoid SVM",
    "Accuracy": accuracy_score(y_test, sigmoid_preds),
    "F1_Positive": f1_score(y_test, sigmoid_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, sigmoid_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, sigmoid_preds, pos_label=1),
    "Training_Time": train_time
})
```

### FAISS Classifier

```python
import faiss
from sklearn.utils.class_weight import compute_class_weight
from sklearn.decomposition import PCA

print("Training FAISS Classifier...")
start = time.time()

# Convert to float32 + contiguous arrays
X_train_np = np.ascontiguousarray(X_train.astype(np.float32))
X_test_np  = np.ascontiguousarray(X_test.astype(np.float32))
y_train_np = y_train.to_numpy().astype(int)

# --------------------------
# PCA for dimensionality reduction
# --------------------------
PCA_DIM = 16
pca = PCA(n_components=PCA_DIM, random_state=42)

X_train_pca = np.ascontiguousarray(pca.fit_transform(X_train_np).astype(np.float32))
X_test_pca  = np.ascontiguousarray(pca.transform(X_test_np).astype(np.float32))

# Normalize for cosine similarity
faiss.normalize_L2(X_train_pca)
faiss.normalize_L2(X_test_pca)

# --------------------------
# Class weights for imbalance
# --------------------------
class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.array([0, 1]),
    y=y_train_np
)
w0, w1 = class_weights

# --------------------------
# Build FAISS Index (HNSW)
# --------------------------
d = X_train_pca.shape[1]
K = 20
M = 32

index = faiss.IndexHNSWFlat(d, M)
index.hnsw.efConstruction = 200
index.hnsw.efSearch = 1024
index.add(X_train_pca)

# --------------------------
# Probability scoring function
# --------------------------
def get_probs():
    distances, neighbors = index.search(X_test_pca, K)
    probs = []

    for dist_list, idx_list in zip(distances, neighbors):
        sim = 1 - dist_list / 2    # cosine → distance → similarity
        sim = np.maximum(sim, 1e-6)

        score0 = 0.0
        score1 = 0.0

        for lbl, s in zip(y_train_np[idx_list], sim):
            if lbl == 0:
                score0 += s * w0
            else:
                score1 += s * w1

        p1 = score1 / (score1 + score0 + 1e-9)
        probs.append(p1)

    return np.array(probs)

# Compute probabilities
probs = get_probs()

# --------------------------
# Threshold tuning
# --------------------------
best_f1 = 0
best_thresh = 0.5
y_test_np = y_test.to_numpy()

for t in np.linspace(0.05, 0.95, 50):
    preds = (probs >= t).astype(int)
    f1 = f1_score(y_test_np, preds)
    if f1 > best_f1:
        best_f1 = f1
        best_thresh = t

# Final predictions
final_preds = (probs >= best_thresh).astype(int)

train_time = time.time() - start

print("Computation Time:", train_time)
print("Best Threshold:", best_thresh)
print("Max F1:", best_f1)
print("\nAccuracy:", accuracy_score(y_test_np, final_preds))
print("\nClassification Report:")
print(classification_report(y_test_np, final_preds))
# ---- Save all FAISS components in one dictionary ----

faiss_info = {
    "pca": pca,
    "index": index,
    "class_weights": class_weights,
    "best_threshold": best_thresh,
    "K": K,
    "PCA_DIM": PCA_DIM
}

print("FAISS info dictionary created.")

# --------------------------
# Log results
# --------------------------
results.append({
    "Model": "FAISS Classifier",
    "Accuracy": accuracy_score(y_test_np, final_preds),
    "F1_Positive": f1_score(y_test_np, final_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test_np, final_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test_np, final_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Baseline Neural Network (MLPClassifier --Adam Solver)

```python
### Baseline Neural Network (MLPClassifier — Adam)

print("Training Baseline Neural Network...")
start = time.time()

nn1 = MLPClassifier(
    hidden_layer_sizes=(128, 64),
    activation="relu",
    solver="adam",
    learning_rate="adaptive",
    max_iter=500,
    random_state=42
)

nn1.fit(X_train, y_train)
train_time = time.time() - start

nn1_preds = nn1.predict(X_test)

print("Training time:", train_time)
print("Accuracy:", accuracy_score(y_test, nn1_preds))
print("\nClassification Report:")
print(classification_report(y_test, nn1_preds))

# Log into results table
results.append({
    "Model": "NeuralNet_Baseline_Adam",
    "Accuracy": accuracy_score(y_test, nn1_preds),
    "F1_Positive": f1_score(y_test, nn1_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, nn1_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, nn1_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Stable Neural Network (LBFGS Solver + RandomOverSampler)

```python
print("Training Stable-Converging NN (LBFGS)...")
start = time.time()

# Oversampling to balance classes
ros = RandomOverSampler(random_state=42)
X_train_res, y_train_res = ros.fit_resample(X_train, y_train)

print("Before oversampling:\n", y_train.value_counts())
print("After oversampling:\n", y_train_res.value_counts(), "\n")

# LBFGS — deterministic & stable convergence
nn_stable = MLPClassifier(
    hidden_layer_sizes=(64, 32, 16),
    activation="relu",
    solver="lbfgs",
    alpha=1e-3,
    max_iter=600,
    random_state=42
)

nn_stable.fit(X_train_res, y_train_res)
train_time = time.time() - start

stable_preds = nn_stable.predict(X_test)

print("Training time:", train_time)
print("Accuracy:", accuracy_score(y_test, stable_preds))
print("\nClassification Report:")
print(classification_report(y_test, stable_preds))

# Log into results table
results.append({
    "Model": "NeuralNet_LBFGS_Oversampled",
    "Accuracy": accuracy_score(y_test, stable_preds),
    "F1_Positive": f1_score(y_test, stable_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, stable_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, stable_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Neural Network — SMOTE + Threshold Tuning

```python
print("Applying SMOTE...")
sm = SMOTE(random_state=42)
X_train_sm, y_train_sm = sm.fit_resample(X_train, y_train)

print("Before SMOTE:\n", y_train.value_counts())
print("After SMOTE:\n", y_train_sm.value_counts(), "\n")

# -------------------------
# Train NN
# -------------------------
print("Training NN on SMOTE data...")
start = time.time()

nn_smote = MLPClassifier(
    hidden_layer_sizes=(64, 32, 16),
    activation="relu",
    solver="adam",
    learning_rate="adaptive",
    learning_rate_init=0.001,
    max_iter=400,
    alpha=1e-4,
    random_state=42
)

nn_smote.fit(X_train_sm, y_train_sm)
train_time = time.time() - start

# -------------------------
# Threshold tuning
# -------------------------
print("\nRunning threshold tuning...")
probs = nn_smote.predict_proba(X_test)[:, 1]

best_f1 = 0
best_t = 0.5

for t in np.arange(0.05, 0.95, 0.01):
    preds_t = (probs >= t).astype(int)
    f1 = f1_score(y_test, preds_t)

    if f1 > best_f1:
        best_f1 = f1
        best_t = t

print("Best Threshold:", round(best_t, 3))
print("Best F1-score:", round(best_f1, 4))

# -------------------------
# Final predictions
# -------------------------
final_preds = (probs >= best_t).astype(int)

print("\nFinal Accuracy:", accuracy_score(y_test, final_preds))
print("\nClassification Report:")
print(classification_report(y_test, final_preds))

# Log into results table
results.append({
    "Model": "NeuralNet_SMOTE_ThresholdTuned",
    "Accuracy": accuracy_score(y_test, final_preds),
    "F1_Positive": f1_score(y_test, final_preds, pos_label=1),
    "Recall_Positive": recall_score(y_test, final_preds, pos_label=1),
    "Precision_Positive": precision_score(y_test, final_preds, pos_label=1),
    "Training_Time": train_time
})
```

### Summary Table of All Model Performances


```python
results_df = pd.DataFrame(results)

# Sort models by F1 score on the positive class (most important metric)
results_df = results_df.sort_values(by="F1_Positive", ascending=False)

results_df.reset_index(drop=True, inplace=True)

results_df
```

### Plot Model Performance (Accuracy, F1, Recall, Precision)

```python
metrics = ["Accuracy", "F1_Positive", "Recall_Positive", "Precision_Positive"]

plt.figure(figsize=(14, 8))

for i, metric in enumerate(metrics):
    plt.subplot(2, 2, i+1)
    plt.barh(results_df["Model"], results_df[metric], color="skyblue")
    plt.title(metric)
    plt.xlabel(metric)
    plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
