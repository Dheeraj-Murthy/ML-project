### Import Libraries

```python
import pandas as pd
import joblib

from sklearn.svm import SVC, LinearSVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report
)

from sklearn.model_selection import GridSearchCV, train_test_split
```

### Load Preprocessed

```python
X_train, X_val, y_train, y_val = joblib.load("../models/processed_train_val.joblib")
X_test = joblib.load("../models/processed_test.joblib")
preprocessor = joblib.load("../models/preprocessor.joblib")

#Load Raw Train for Resampling 20%
train_df = pd.read_csv("../data/train_updated.csv")
X_raw_full = train_df.drop("RiskFlag", axis=1)
y_raw_full = train_df["RiskFlag"]
```

### Evaluation Function

```python
def evaluate(name, y_true, y_pred, y_prob=None):
    print(f"\n===== {name} =====")
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred))
    print("Recall   :", recall_score(y_true, y_pred))
    print("F1-score :", f1_score(y_true, y_pred))
    if y_prob is not None:
        print("ROC-AUC :", roc_auc_score(y_true, y_prob))
    print("\nConfusion:\n", confusion_matrix(y_true, y_pred))
    print("\nReport:\n", classification_report(y_true, y_pred))
```

### Baseline Linear SVM

```python
linear_full = LinearSVC(
    class_weight="balanced",
    C=1.0,
    max_iter=10000
)

linear_full.fit(X_train, y_train)

pred_lin_full = linear_full.predict(X_val)
evaluate("Linear SVM — FULL Baseline", y_val, pred_lin_full)

f1_lin_full_base = f1_score(y_val, pred_lin_full)
```

### Hyperparameter Tuning

```python
param_lin = {
    "C": [0.01, 0.1, 1, 5, 10]
}

linear_grid_full = GridSearchCV(
    LinearSVC(
        class_weight="balanced",
        max_iter=15000
    ),
    param_grid=param_lin,
    scoring="f1",
    cv=3,
    verbose=1,
    n_jobs=-1
)

linear_grid_full.fit(X_train, y_train)
print("Best Linear FULL Params:", linear_grid_full.best_params_)

best_linear_full = linear_grid_full.best_estimator_

pred_lin_full_tuned = best_linear_full.predict(X_val)
evaluate("Linear SVM — FULL Tuned", y_val, pred_lin_full_tuned)

f1_lin_full_tuned = f1_score(y_val, pred_lin_full_tuned)
```

### Baseline RBF Kernel

```python
rbf_full = SVC(
    kernel="rbf",
    C=1,
    gamma="scale",
    probability=True,
    class_weight="balanced"
)

rbf_full.fit(X_train, y_train)

pred_rbf = rbf_full.predict(X_val)
prob_rbf = rbf_full.predict_proba(X_val)[:,1]

evaluate("RBF SVM — FULL Baseline", y_val, pred_rbf, prob_rbf)
f1_rbf_full_base = f1_score(y_val, pred_rbf)
```

### Tuned RBF

```python
param_rbf = {
    "C": [1, 5, 10],
    "gamma": ["scale", 0.01, 0.001]
}

rbf_grid_full = GridSearchCV(
    SVC(
        kernel="rbf",
        probability=True,
        class_weight="balanced"
    ),
    param_grid=param_rbf,
    scoring="f1",
    cv=3,
    verbose=1,
    n_jobs=-1
)

rbf_grid_full.fit(X_train, y_train)
print("Best RBF FULL Params:", rbf_grid_full.best_params_)

best_rbf_full = rbf_grid_full.best_estimator_

pred_rbf_full_tuned = best_rbf_full.predict(X_val)
prob_rbf_full_tuned = best_rbf_full.predict_proba(X_val)[:,1]

evaluate("RBF SVM — FULL Tuned", y_val, pred_rbf_full_tuned, prob_rbf_full_tuned)

f1_rbf_full_tuned = f1_score(y_val, pred_rbf_full_tuned)
```

### 20% subset

```python
df_20 = train_df.sample(frac=0.2, random_state=42)
X_raw_20 = df_20.drop("RiskFlag", axis=1)
y_raw_20 = df_20["RiskFlag"]

X_20 = preprocessor.transform(X_raw_20)

X_train_20, X_val_20, y_train_20, y_val_20 = train_test_split(
    X_20, y_raw_20,
    test_size=0.2,
    random_state=42,
    stratify=y_raw_20
)
```

### Baseline Linear SVM - 20%

```python
linear_20 = LinearSVC(
    class_weight="balanced",
    C=1,
    max_iter=10000
)

linear_20.fit(X_train_20, y_train_20)

pred_lin_20 = linear_20.predict(X_val_20)
evaluate("Linear SVM — 20% Baseline", y_val_20, pred_lin_20)

f1_lin_20_base = f1_score(y_val_20, pred_lin_20)
```

### Parameter Tuning

```python
linear_grid_20 = GridSearchCV(
    LinearSVC(
        class_weight="balanced",
        max_iter=15000
    ),
    param_grid=param_lin,
    scoring='f1',
    cv=3,
    verbose=1,
    n_jobs=-1
)

linear_grid_20.fit(X_train_20, y_train_20)
print("Best Linear 20% Params:", linear_grid_20.best_params_)

best_linear_20 = linear_grid_20.best_estimator_

pred_lin_20_tuned = best_linear_20.predict(X_val_20)
evaluate("Linear SVM — 20% Tuned", y_val_20, pred_lin_20_tuned)

f1_lin_20_tuned = f1_score(y_val_20, pred_lin_20_tuned)
```

### RBF SVM (20%)

```python
rbf_20 = SVC(
    kernel="rbf",
    C=1,
    gamma="scale",
    class_weight="balanced",
    probability=True
)

rbf_20.fit(X_train_20, y_train_20)

pred_rbf_20 = rbf_20.predict(X_val_20)
prob_rbf_20 = rbf_20.predict_proba(X_val_20)[:, 1]

evaluate("RBF SVM — 20% Baseline", y_val_20, pred_rbf_20, prob_rbf_20)
f1_rbf_20_base = f1_score(y_val_20, pred_rbf_20)
```

### Tune RBF

```python
rbf_grid_20 = GridSearchCV(
    SVC(
        kernel="rbf",
        probability=True,
        class_weight="balanced"
    ),
    param_grid=param_rbf,
    scoring="f1",
    cv=3,
    n_jobs=-1,
    verbose=1
)

rbf_grid_20.fit(X_train_20, y_train_20)
print("Best RBF Params 20%:", rbf_grid_20.best_params_)
best_rbf_20 = rbf_grid_20.best_estimator_

pred_rbf_20_tuned = best_rbf_20.predict(X_val_20)
prob_rbf_20_tuned = best_rbf_20.predict_proba(X_val_20)[:, 1]

evaluate("RBF SVM — 20% Tuned", y_val_20, pred_rbf_20_tuned, prob_rbf_20_tuned)
f1_rbf_20_tuned = f1_score(y_val_20, pred_rbf_20_tuned)
```

### Result Table

```python
results_svm = pd.DataFrame({
    "Model": [
        "Linear FULL Baseline",
        "Linear FULL Tuned",
        "RBF FULL Baseline",
        "RBF FULL Tuned",
        "Linear 20% Baseline",
        "Linear 20% Tuned",
        "RBF 20% Baseline",
        "RBF 20% Tuned"
    ],
    "F1 Score": [
        f1_lin_full_base,
        f1_lin_full_tuned,
        f1_rbf_full_base,
        f1_rbf_full_tuned,
        f1_lin_20_base,
        f1_lin_20_tuned,
        f1_rbf_20_base,
        f1_rbf_20_tuned
    ]
})

results_svm
```

### Final Training for submissions

```python
X_full_pre = preprocessor.transform(X_raw_full)

final_linear_full = best_linear_full
final_rbf_full = best_rbf_full

final_linear_full.fit(X_full_pre, y_raw_full)
final_rbf_full.fit(X_full_pre, y_raw_full)

pred_linear_full = final_linear_full.predict(X_test)
pred_rbf_full = final_rbf_full.predict(X_test)

sub_lin_full = pd.DataFrame({
    "ProfileID": train_df["ProfileID"].iloc[:len(X_test)],  # if needed
    "RiskFlag": pred_linear_full
})
sub_lin_full.to_csv("s../submissions/ubmission_svm_linear_full.csv", index=False)

sub_rbf_full = pd.DataFrame({
    "ProfileID": train_df["ProfileID"].iloc[:len(X_test)],
    "RiskFlag": pred_rbf_full
})
sub_rbf_full.to_csv("../submissions/submission_svm_rbf_full.csv", index=False)

print("Full SVM submissions saved")

X_20_pre = preprocessor.transform(X_raw_20)

final_linear_20 = best_linear_20
final_rbf_20 = best_rbf_20

final_linear_20.fit(X_20_pre, y_raw_20)
final_rbf_20.fit(X_20_pre, y_raw_20)

pred_linear_20 = final_linear_20.predict(X_test)
pred_rbf_20 = final_rbf_20.predict(X_test)

pd.DataFrame({
    "ProfileID": train_df["ProfileID"].iloc[:len(X_test)],
    "RiskFlag": pred_linear_20
}).to_csv("../submissions/submission_svm_linear_20.csv", index=False)

pd.DataFrame({
    "ProfileID": train_df["ProfileID"].iloc[:len(X_test)],
    "RiskFlag": pred_rbf_20
}).to_csv("../submissions/submission_svm_rbf_20.csv", index=False)

print("20% SVM submissions saved")

joblib.dump(final_linear_full, "../models/svm_linear_best_full.joblib")
joblib.dump(final_rbf_full, "../models/svm_rbf_best_full.joblib")

joblib.dump(final_linear_20, "../models/svm_linear_best_20.joblib")
joblib.dump(final_rbf_20, "../models/svm_rbf_best_20.joblib")

print("All models saved")
```
